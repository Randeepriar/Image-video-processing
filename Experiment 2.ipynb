Case 1: Convert the RGB image to grayscale by taking the mean average of the three color planes (R, G, and B).
Case 2: Input the weightages of the three color planes (R, G, and B) from the user. The weightage of each plane should be a value between 0 and 1, and the sum of all three weightages must be equal to 1.
Example:
Weightages of R, G, and B are 0.7, 0.2, and 0.1, respectively.

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read image in grayscale
image = cv2.imread("input.jpg", cv2.IMREAD_GRAYSCALE)

# Calculate mean intensity
threshold = np.mean(image)
print("Threshold (Mean Intensity):", threshold)

# Convert to binary
_, binary = cv2.threshold(image, threshold, 255, cv2.THRESH_BINARY)

# Display images
print("Grayscale Image")
cv2_imshow(image)

print("Binary Image")
cv2_imshow(binary)

# Save output
cv2.imwrite("binary_mean.jpg", binary)
```

```python
import cv2
from google.colab.patches import cv2_imshow

# Read image in grayscale
image = cv2.imread("input.jpg", cv2.IMREAD_GRAYSCALE)

# User input
threshold = int(input("Enter threshold value (0-255): "))

# Convert to binary
_, binary = cv2.threshold(image, threshold, 255, cv2.THRESH_BINARY)

# Display images
print("Grayscale Image")
cv2_imshow(image)

print("Binary Image")
cv2_imshow(binary)

# Save output
cv2.imwrite("binary_user.jpg", binary)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read color image
image = cv2.imread("input.jpg")

# Split into B, G, R channels
B, G, R = cv2.split(image)

# Mean average
gray = ((R.astype(float) + G.astype(float) + B.astype(float)) / 3).astype(np.uint8)

# Display
print("Original Image")
cv2_imshow(image)

print("Grayscale Image")
cv2_imshow(gray)

# Save
cv2.imwrite("gray_average.jpg", gray)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read color image
image = cv2.imread("input.jpg")

# Split channels
B, G, R = cv2.split(image)

# Input weights
wr = float(input("Enter weight for Red (0-1): "))
wg = float(input("Enter weight for Green (0-1): "))
wb = float(input("Enter weight for Blue (0-1): "))

# Check validity
if abs((wr + wg + wb) - 1) != 1:
    print("Error: Sum of weights must be 1.")
else:
    gray = (wr * R + wg * G + wb * B).astype(np.uint8)

    print("Original Image")
    cv2_imshow(image)

    print("Grayscale Image")
    cv2_imshow(gray)

    cv2.imwrite("gray_weighted.jpg", gray)
    print("Grayscale image saved successfully.")
```
