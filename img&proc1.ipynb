```python
import cv2
from google.colab.patches import cv2_imshow

# Read image
image = cv2.imread("input.jpg")

# Display image
cv2_imshow(image)

# Save image
cv2.imwrite("output.jpg", image)

print("Image read, displayed, and saved successfully.")
```

```python
import cv2

image = cv2.imread("input.jpg")

cv2.imwrite("image.png", image)
cv2.imwrite("image.tiff", image)
cv2.imwrite("image.bmp", image)

print("Image converted into PNG, TIFF, and BMP.")
```

```python
import cv2
from google.colab.patches import cv2_imshow

image = cv2.imread("input.jpg")

resized = cv2.resize(image, (200, 100))

cv2_imshow(resized)

cv2.imwrite("resized.jpg", resized)

print("Image resized successfully.")
```

```python
import cv2
from google.colab.patches import cv2_imshow

image = cv2.imread("input.jpg")

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

cv2_imshow(gray)

cv2.imwrite("grayscale.jpg", gray)

print("Grayscale image saved successfully.")
```

```python
import cv2
from google.colab.patches import cv2_imshow

video = cv2.VideoCapture("vidinput.mp4")

frame = 0

while True:
    ret, img = video.read()

    if not ret:
        break

    print("Frame:", frame)
    cv2_imshow(img)

    frame += 1

    if frame == 10:
        break

video.release()
```

```python
import cv2
import os

video = cv2.VideoCapture("vidinput.mp4")

# Create folder to save frames
os.makedirs("frames", exist_ok=True)

count = 0

while True:
    ret, frame = video.read()

    if not ret:
        break

    cv2.imwrite(f"frames/frame_{count}.jpg", frame)
    count += 1

video.release()

print("Total Frames Extracted:", count)
print("Frames saved in the 'frames' folder.")
```
