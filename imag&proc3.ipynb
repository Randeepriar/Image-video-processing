```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()
```

```python
filename = list(uploaded.keys())[0]

img = cv2.imread(filename)

print("Image loaded successfully!")
print("Image size:", img.shape)
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 5))
plt.imshow(img_rgb)
plt.axis('off')
plt.title('Original Image')
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(6, 5))
plt.imshow(gray, cmap='gray')
plt.axis('off')
plt.title('Grayscale Image')
plt.show()
```

```python
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(6, 5))
plt.imshow(binary, cmap='gray')
plt.axis('off')
plt.title('Binary Image')
plt.show()
```

```python
border_width = int(input("Enter border width in pixels: "))

print("Enter border color:")
print("0 = Black")
print("1 = White")

color_choice = int(input("Enter your choice: "))

if color_choice == 0:
    border_color = 0
else:
    border_color = 255
```

```python
gray_border = cv2.copyMakeBorder(
    gray,
    border_width,
    border_width,
    border_width,
    border_width,
    cv2.BORDER_CONSTANT,
    value=border_color
)

plt.figure(figsize=(6, 5))
plt.imshow(gray_border, cmap='gray')
plt.axis('off')
plt.title('Grayscale Image with Border')
plt.show()
```

```python
binary_border = cv2.copyMakeBorder(
    binary,
    border_width,
    border_width,
    border_width,
    border_width,
    cv2.BORDER_CONSTANT,
    value=border_color
)

plt.figure(figsize=(6, 5))
plt.imshow(binary_border, cmap='gray')
plt.axis('off')
plt.title('Binary Image with Border')
plt.show()
```

```python
gray_complement = 255 - gray

plt.figure(figsize=(6, 5))
plt.imshow(gray_complement, cmap='gray')
plt.axis('off')
plt.title('Complement of Grayscale Image')
plt.show()
```

```python
binary_complement = 255 - binary

plt.figure(figsize=(6, 5))
plt.imshow(binary_complement, cmap='gray')
plt.axis('off')
plt.title('Complement of Binary Image')
plt.show()
```

```python
cv2.imwrite('/content/grayscale_border.jpg', gray_border)
cv2.imwrite('/content/binary_border.jpg', binary_border)
cv2.imwrite('/content/grayscale_complement.jpg', gray_complement)
cv2.imwrite('/content/binary_complement.jpg', binary_complement)

print("All output images saved successfully!")
```
